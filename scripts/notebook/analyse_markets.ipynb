```python
import json
import pandas as pd


def wal_to_frames(path: str) -> dict[str, pd.DataFrame]:
    """一个 WAL jsonl 文件 -> {"price_change": df, "book": df, "trade": df}
    price_change: 每个 asset 一行 (一条报文含两个互补 token, 所以行数 = 报文数 x 2)
    book:         每个价位一行 (bids/asks 摊平)
    trade:        每条报文一行"""
    pc, book, trade = [], [], []
    with open(path) as f:
        for line in f:
            if not line.strip():
                continue
            #rec: dict
            rec = json.loads(line)
            m = json.loads(rec["raw"])  # raw 是二次编码的字符串, 要再解一次
            if not isinstance(m, dict):  # 初始订阅回包是 list, 跳过
                m = m[0]
                for side, levels in (("bid", m["bids"]), ("ask", m["asks"])):
                    for lv in levels:
                        book.append({**base, "asset_id": m["asset_id"], "side": side,
                                     "price": lv["price"], "size": lv["size"]})

            base = {"ts_recv_ns": rec["ts_recv_ns"],
                    "ts_server_ms": int(m["timestamp"]),
                    "market": m.get("market")}
            et = m.get("event_type")
            if et == "price_change":
                for ch in m["price_changes"]:
                    pc.append({**base, **ch})
            elif et == "book":
                for side, levels in (("bid", m["bids"]), ("ask", m["asks"])):
                    for lv in levels:
                        book.append({**base, "asset_id": m["asset_id"], "side": side,
                                     "price": lv["price"], "size": lv["size"]})
            elif et == "last_trade_price":
                trade.append({**base, "asset_id": m["asset_id"], "price": m["price"],
                              "size": m["size"], "side": m["side"],
                              "fee_rate_bps": m.get("fee_rate_bps")})
    out = {}
    for name, rows in (("price_change", pc), ("book", book), ("trade", trade)):
        df = pd.DataFrame(rows)
        if not df.empty:
            for col in ("price", "size", "best_bid", "best_ask"):
                if col in df:
                    df[col] = pd.to_numeric(df[col])
            df["ts_recv"] = pd.to_datetime(df["ts_recv_ns"], unit="ns", utc=True)
            df["ts_server"] = pd.to_datetime(df["ts_server_ms"], unit="ms", utc=True)
            df["lag_ms"] = df["ts_recv_ns"] / 1e6 - df["ts_server_ms"]
        out[name] = df
    return out


frames = wal_to_frames("../../data_bk/orchestrator/wal/2026-09-14/gen0/5.jsonl")
pc, book, trade = frames["price_change"], frames["book"], frames["trade"]
```
